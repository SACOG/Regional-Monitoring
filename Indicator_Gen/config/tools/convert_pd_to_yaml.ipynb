```python


# Packages
import pandas as pd
import numpy as np
import yaml

import os
import getpass
from pathlib import Path


# File paths
user = getpass.getuser()
path_users = Path.home()
path_git = path_users / 'Documents' / 'Projects' / 'Regional-Monitoring' / 'Indicator_Gen'
path_code    = path_git / 'Data' / 'BLS'
path_config0 = path_git / 'config'
path_config  = path_code / 'config'

```

```python


file_area_codes = path_config0 / 'area_codes.xlsx'


df_states = pd.read_excel(file_area_codes, sheet_name='StateNames', dtype=str)
display(df_states.head())


list_states = df_states.to_dict(orient='records')

yaml_states = {}

for state in list_states:
    yaml_states[state['State']] = state

display(yaml_states)


file_yaml = path_config0 / 'state_codes.yaml'
with open(file_yaml, 'w') as file:
    yaml.dump(yaml_states, file, default_flow_style=False)
    print("YAML file 'state_codes.yaml' created successfully.")


```
