```python


import numpy as np
import pandas as pd
from pathlib import Path
import plotly.express as px
from IPython.display import display
import sys
sys.path.append(str(Path.home()/'Documents'/'Projects'/'Regional-Monitoring'/'Indicator_Gen'/'config'))
import functions as func
pd.set_option('display.max_columns', None)

PATH_SP = Path.home() / 'Sacramento Area Council of Governments' / 'Regional Monitoring and Reporting - Documents' / 'Products' / 'Chamber Study Missions' / '2026'
PATH_I = Path(r'I:\Projects\Josh\Regional Monitoring\Task 9. Collect new data\BEA')


EXPORT=False

```

```python

df1 = pd.read_csv(PATH_I/'CAGDP9'/'CAGDP9_CA_2001_2024.csv', na_values="(D)")
df1 = df1[df1['GeoName'].isin(['El Dorado, CA', 'Placer, CA', 'Sacramento, CA', 'Sutter, CA', 'Yolo, CA', 'Yuba, CA'])]
df1['MPO'] = 'SACOG'

df2 = pd.read_csv(PATH_I/'CAGDP9'/'CAGDP9_MI_2001_2024.csv', na_values="(D)")
df2 = df2[df2['GeoName'].isin(['Livingtson, MI', 'Oakland, MI', 'Macomb, MI', 'St. Clair, MI', 'Wayne, MI', 'Washtenaw, MI', 'Monroe, MI'])]
df2['MPO'] = 'SEMCOG'

df = pd.concat([df1, df2])
df = df.melt(id_vars=['MPO', 'GeoFIPS', 'GeoName', 'Region', 'TableName', 'LineCode', 'IndustryClassification', 'Description', 'Unit'], var_name='Year', value_name='GRP (Thousands of Dollars)')
df = df[df['Description']=='All industry total ']
df['GRP (Thousands of Dollars)'] = df['GRP (Thousands of Dollars)'].astype(int)
df['Year'] = df['Year'].astype(int)
df = df.groupby(['MPO', 'Year'], as_index=False)['GRP (Thousands of Dollars)'].sum()

df = df.sort_values(['MPO', 'Year'], ascending = [True, True]).reset_index(drop=True)

df['AnnualPctDiff'] = df['GRP (Thousands of Dollars)'].pct_change()
df.loc[df['Year'] == 2001, 'AnnualPctDiff'] = np.nan
df.loc[df['AnnualPctDiff'] == np.inf, 'AnnualPctDiff'] = np.nan

year_max = np.max(df['Year'].unique())
year_min = np.min(df['Year'].unique())

col_pct_change = f'PctChange_{year_min}_{year_max}'
col_annual_pct_change = f'AnnualPctDiff_{year_min}_{year_max}'

df_summary = df.copy()
df_summary = df_summary[(df_summary['Year'] == year_max) | (df_summary['Year'] == year_min)]
df_summary = df_summary.drop('AnnualPctDiff', axis = 1)
df_summary[col_pct_change] = df_summary['GRP (Thousands of Dollars)'].pct_change()
df_summary.loc[df_summary['Year'] == year_min, col_pct_change] = np.nan
df_summary.loc[df_summary[col_pct_change] == np.inf, col_pct_change] = np.nan
df_summary = df_summary[df_summary['Year'] == year_max]
df_summary = df_summary.drop(['Year', 'GRP (Thousands of Dollars)'], axis=1)

df_wm = df.copy()
df_wm = df_wm.groupby(['MPO'], as_index = False).agg(weighted_mean_grp = ('AnnualPctDiff', 'mean'))
df_wm = df_wm.rename(columns={'weighted_mean_grp':col_annual_pct_change})

df_summary = df_summary.merge(df_wm, on = 'MPO')

df         = df        .sort_values(['MPO', 'Year'], ascending=[True, False]).reset_index(drop=True)
df_summary = df_summary.sort_values(['MPO'        ], ascending=[True       ]).reset_index(drop=True)

df
```
