```python


import pandas as pd
from pathlib import Path
import plotly.express as px

PATH_SP = Path.home() / 'Sacramento Area Council of Governments' / 'Regional Monitoring and Reporting - Documents' / 'Products' / 'Chamber Study Missions' / '2026' / 'Indicators'
PATH_I = Path(r'I:\Projects\Josh\Regional Monitoring')

```

```python
indicator='Pop_3'
df = pd.read_excel(PATH_SP/f'{indicator} MPO ACS5_prt_SEMCOG.xlsx', sheet_name='MPO')
px.bar(df[(df['Year']==2024) & (df['Race/Ethnicity']!='All')], x='Race/Ethnicity', y='Percent', color='MPO', barmode='group')
```

```python
indicator='Pop_3'
df = pd.read_excel(PATH_SP/f'{indicator} MPO ACS5_prt_SEMCOG.xlsx', sheet_name='PctPopGrowth')
df = df.melt(id_vars=['MPO', 'Year'], var_name='Race/Ethnicity', value_name='PctChange')
px.bar(df, x='Race/Ethnicity', y='PctChange', color='MPO', barmode='group')
```

```python
indicators = ['Pop_4', 'Commute_1']

for indicator in indicators:
    df = pd.read_excel(PATH_SP/f'{indicator} MPO ACS5_prt_SEMCOG.xlsx', sheet_name='MPO')
    fig = px.bar(df, x='Variable', y='Percent', color='MPO', barmode='group')
    fig.update_layout(title=indicator)
    fig.show()
```

```python

indicators = ['Edu_1', 'Cost_5', 'Income_4']

for indicator in indicators:
    df = pd.read_excel(PATH_SP/f'{indicator} MPO ACS5_prt_SEMCOG.xlsx', sheet_name='MPO')
    fig = px.bar(df[df['Race/Ethnicity']=='White (NH)'], x='Variable', y='Percent', color='MPO', barmode='group')
    fig.update_layout(title=indicator)
    fig.show()
```

```python
indicator = 'Income_1'
df = pd.read_excel(PATH_SP/f'{indicator} MPO ACS5_prt_SEMCOG.xlsx', sheet_name='MPO')
px.bar(df, x='Race/Ethnicity', y='Median Household Income', color='MPO', barmode='group')
```

```python

indicators = ['Cost_1', 'Cost_2']

for indicator in indicators:
    df = pd.read_excel(PATH_SP/f'{indicator} Region Zillow_prt_SEMCOG.xlsx', sheet_name='Annual')
    df.columns = [col.lower() for col in df.columns]

    x = 'year'
    y = 'price'
    color = 'msa'
    labels = 'msa'

    fig = px.line(df, x = x, y = y, color = color, markers = True, labels = labels)
    fig.update_layout(title=indicator)
    fig.show()
```

```python

indicator='Jobs_1'
df = pd.read_excel(PATH_SP/f'{indicator} MSA BLS SM_prt_SEMCOG.xlsx', sheet_name='MSA')

sectors = ['All', 'Total Private', 'Government']
for sector in sectors:
    fig = px.line(df[df['Sector']==sector], x='date_', y='Total Jobs', color='MSA')
    fig.update_layout(title=sector)
    fig.show()
```
