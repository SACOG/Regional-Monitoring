```python


EXPORT=True

from pathlib import Path
import pandas as pd

PATH_DATA = Path(r'\\webmapping-svr\c$\inetpub\wwwroot\monitoring\Data')
PATH_SDR = Path(r'C:\Users\jfontes\Sacramento Area Council of Governments\Regional Monitoring and Reporting - Documents\Products\Small Data Requests\2025\SacEJ Pla')


df_cost3 = pd.read_excel(PATH_DATA / 'Cost_3 Block Groups ACS5.xlsx', sheet_name='Block Groups')
df_cost3 = df_cost3[df_cost3['County Name'] == 'Placer'].dropna().reset_index(drop=True)
df_cost3 = df_cost3.drop(['State FIPS', 'County FIPS', 'Tract ID', 'Block Group ID', 'Race_Ethnicity'], axis=1)
display(df_cost3)


df_cost1 = pd.read_excel(PATH_DATA / 'Cost_1 ZIP Codes Sales Price Zillow.xlsx', sheet_name='Annual')
df_cost1 = df_cost1[df_cost1['CountyName'] == 'Placer County'].dropna().reset_index(drop=True)
df_cost1_zip = df_cost1.copy()
display(df_cost1_zip)


df_cost1 = pd.read_excel(PATH_DATA / 'Cost_1 Neighborhoods Sales Price Zillow.xlsx', sheet_name='Annual')
df_cost1 = df_cost1[df_cost1['CountyName'] == 'Placer County'].dropna().reset_index(drop=True)
df_cost1_neighborhoods = df_cost1.copy()
display(df_cost1_neighborhoods)


df_cost1 = pd.read_excel(PATH_DATA / 'Cost_1 Cities Sales Price Zillow.xlsx', sheet_name='Annual')
df_cost1 = df_cost1[df_cost1['CountyName'] == 'Placer County'].dropna().reset_index(drop=True)
df_cost1_cities = df_cost1.copy()
display(df_cost1_cities)


df_cost2 = pd.read_excel(PATH_DATA / 'Cost_2 ZIP Codes Rent Price Zillow.xlsx', sheet_name='Annual')
df_cost2 = df_cost2[df_cost2['CountyName'] == 'Placer County'].dropna().reset_index(drop=True)
df_cost2_zip = df_cost2.copy()
display(df_cost2_zip)


df_cost2 = pd.read_excel(PATH_DATA / 'Cost_2 Cities Rent Price Zillow.xlsx', sheet_name='Annual')
df_cost2 = df_cost2[df_cost2['CountyName'] == 'Placer County'].dropna().reset_index(drop=True)
df_cost2_cities = df_cost2.copy()
display(df_cost2_cities)


if EXPORT:
    with pd.ExcelWriter(PATH_SDR / 'Placer_County_Cost.xlsx', engine='xlsxwriter') as writer:
        df_cost3              .to_excel(writer, index=False, sheet_name='BlockGroups_Vacancies')
        df_cost1_zip          .to_excel(writer, index=False, sheet_name='Sales_ZIP'            )
        df_cost1_neighborhoods.to_excel(writer, index=False, sheet_name='Sales_Neighborhoods'  )
        df_cost1_cities       .to_excel(writer, index=False, sheet_name='Sales_Cities'         )
        df_cost2_zip          .to_excel(writer, index=False, sheet_name='Rent_ZIP'             )
        df_cost2_cities       .to_excel(writer, index=False, sheet_name='Rent_Cities'          )

    
```
