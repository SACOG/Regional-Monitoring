```python


indicator = 'HouseholdIncome_1'
geography = 'MPO'
estimate = 'PUMS1'

wkbook_in = f'{indicator} {geography} {estimate}.xlsx'
file_in = path_sdr / 'orig' / wkbook_in
df_acs_orig = pd.read_excel(file_in)
df_acs_orig

df_acs = df_acs_orig.copy()

df_acs = df_acs[df_acs['MPO'] == 'SACOG']
df_acs['HINCP'] = df_acs['HINCP'].astype('int64')

conditions = [
    df_acs['HINCP'] < 15000,
    (df_acs['HINCP'] >=  15000) & (df_acs['HINCP'] <  25000),
    (df_acs['HINCP'] >=  25000) & (df_acs['HINCP'] <  35000),
    (df_acs['HINCP'] >=  35000) & (df_acs['HINCP'] <  50000),
    (df_acs['HINCP'] >=  50000) & (df_acs['HINCP'] <  75000),
    (df_acs['HINCP'] >=  75000) & (df_acs['HINCP'] < 100000),
    (df_acs['HINCP'] >= 100000) & (df_acs['HINCP'] < 150000),
    (df_acs['HINCP'] >= 150000) & (df_acs['HINCP'] < 200000),
    (df_acs['HINCP'] >= 200000) & (df_acs['HINCP'] < 250000),
    (df_acs['HINCP'] >= 250000)
]

choices = ['Income under 15k', 'Income 15k-25k', 'Income 25k-35k', 'Income 35k-50k', 'Income 50k-75k', 'Income 75k-100k', 'Income 100k-150k', 'Income 150k-200k', 'Income 200k-250k', 'Income over 250k']

df_acs['Household Income Bracket'] = np.select(conditions, choices, default='no')



df_acs = df_acs.groupby(['MPO', 'Year', 'Household Income Bracket'], as_index=False)['Households'].sum()

df_acs['Percentage'] = df_acs['Households'] / df_acs.groupby(['MPO', 'Year'], as_index=False)['Households'].transform('sum')

df_acs['sort'] = pd.Categorical(df_acs['Household Income Bracket'], ['Income under 15k', 'Income 15k-25k', 'Income 25k-35k', 'Income 35k-50k', 'Income 50k-75k', 'Income 75k-100k', 'Income 100k-150k', 'Income 150k-200k', 'Income 200k-250k', 'Income over 250k'])
df_acs = df_acs.sort_values(['Year', 'sort'], ascending=[False, True])
df_acs = df_acs.drop('sort', axis=1)
df_acs = df_acs.reset_index(drop=True)

display(df_acs)

if export:
    wkbook_out = f'{indicator} {geography} {estimate}_clean.xlsx'
    file_out = path_sdr / wkbook_out
    df_acs.to_excel(file_out, index=False)



df_qc = df_acs.groupby(['MPO', 'Year'], as_index=False)['Households'].sum()
display(df_qc)


df_qc = df_acs_orig.copy()

df_qc = df_qc[df_qc['MPO'] == 'SACOG']
df_qc = df_qc.groupby(['MPO', 'Year'], as_index=False)['Households'].sum()

display(df_qc)

```
