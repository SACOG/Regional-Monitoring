```python


import pandas as pd
from pathlib import Path
from IPython.display import display


PATH_SP   = Path.home() / 'Sacramento Area Council of Governments\Regional Monitoring and Reporting - Documents'
PATH_DOF = PATH_SP / 'Data' / 'Vibrant and Inclusive Places' / 'People and Community' / 'Pop and Demographics'
PATH_PROD = PATH_SP / 'Products' / 'Small Data Requests' / '2025' / 'TDA Funding'

FILE_DOF = PATH_DOF / 'DOF_E5_and_E8_Jurisdictions.xlsx'
FILE_OUT = PATH_PROD / 'TDA Funding update.xlsx'


df_tda = pd.read_excel(FILE_DOF)
df_tda = df_tda[df_tda['Year'] == 2025]
df_tda = df_tda[df_tda['County'].isin(['El Dorado', 'Placer', 'Sacramento', 'Sutter', 'Yolo', 'Yuba'])]
df_tda.loc[df_tda['Jurisdiction'] == 'Unincorporated', 'Jurisdiction'] = df_tda['County'] + ' ' + df_tda['Jurisdiction']
df_tda = df_tda[['County', 'Jurisdiction', 'Population']]

jurisdictions = [
    'Placerville', 'South Lake Tahoe', 'El Dorado Unincorporated', 
    'Auburn', 'Colfax', 'Lincoln', 'Loomis Town', 'Rocklin', 'Roseville', 'Placer Unincorporated'
    'Citrus Heights', 'Elk Grove', 'Folsom', 'Galt', 'Isleton', 'Rancho Cordova', 'Sacramento', 'Sacramento Unincorporated',
    'Live Oak', 'Yuba', 'Sutter Uncinorporated',
    'Davis', 'West Sacramento', 'Winters', 'Woodland', 'Yolo Unincorporated',
    'Marysville', 'Wheatland', 'Yuba Unincorporated'
]

df_tda[df_tda['Jurisdiction'].isin(jurisdictions)]


display(df_tda.head())

df_tda.to_excel(FILE_OUT, index=False)

```
