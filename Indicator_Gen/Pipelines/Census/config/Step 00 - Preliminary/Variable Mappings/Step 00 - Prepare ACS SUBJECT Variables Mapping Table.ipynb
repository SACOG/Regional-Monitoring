```python
# General
import numpy as np
import pandas as pd
import os


# Read in json files stored at url
import urllib.request, json 
```

```python
# Define user
user = os.getlogin()

# Working directories
path_sp  = os.path.join('C:\\Users', 'jfontes', 'Sacramento Area Council of Governments\Regional Monitoring and Reporting - Documents')
path_git = os.path.join('C:\\Users', 'jfontes', 'Documents', 'Projects', 'Regional-Monitoring', 'Indicator_Gen')

# Set file paths
path_config = os.path.join(path_git, 'Python Code', 'Census', 'aa_config')
```

```python
with urllib.request.urlopen("https://api.census.gov/data/2022/acs/acs5/subject/variables.json") as url:
    dict_acs = json.load(url)
```

```python
df_vars = pd.DataFrame.from_dict(dict_acs['variables']).T.reset_index().rename(columns = {'index':'ID'})

df_vars = df_vars[['ID', 'attributes', 'label', 'concept']]
df_vars['Label_clean'] = df_vars['label'].str.replace('Estimate!!', '')
df_vars['Label_clean'] = df_vars['Label_clean'].str.replace('!!', ' ')
df_vars['Label_clean'] = df_vars['Label_clean'].str.replace(':', '')

print(df_vars.shape)
df_vars.head()
```

```python
df_vars.to_excel(os.path.join(path_config, 'ACS SUBJECT 5-Year Estimates Variable Tables and Labels 2022 JSON.xlsx'), index=False)
```
