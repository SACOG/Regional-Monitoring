[Revenue Vehicles by Useful Life Benchmark (ULB) by Region](https://catalog.dvrpc.org/dataset/transit-conditions/resource/f6b47c4e-c321-4e1e-998a-2120461e240a)
1) From the NTD vehicles inventory, for each year, agency (SEPTA, PATCO, PART, and NJ Transit), and mode combination, get the number of revenue vehicles which met or exceeded their useful life benchmark (ULB)
2) From the same inventory, for each year, agency (SEPTA, PATCO, PART, and NJ Transit), and mode combination, get the number of revenue vehicles which reported a ULB
3) From the same inventory, for each year, agency (SEPTA, PATCO, PART, and NJ Transit), and mode combination, get the percent of revenue vehicles which met or exceeded their useful life benchmark (ULB) (step 1 divided by step 2)
4) Get the same data as steps 1-3, but for all agencies summed up (therefore, each year and mode combo)
5) Get the same data as steps 1-3, but for all modes summed up (therefore, each year and agency combo)
6) Union the results from steps 1-5
