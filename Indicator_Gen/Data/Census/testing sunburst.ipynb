```python
import pandas as pd
import plotly.express as px
import os

def read_vmt_data(filepath):
    excel_file = pd.ExcelFile(filepath)
    return {sheet: pd.read_excel(filepath, sheet_name=sheet) 
            for sheet in excel_file.sheet_names}

def create_sunburst(data_dict, output_folder):
    ids = []
    parents = []
    values = []
    labels = []
    sort_keys = []  # Add sort keys for ordering
    
    ids.append('Total')
    parents.append('')
    values.append(0)
    labels.append('Total VMT<br>2019-2024')
    sort_keys.append(0)

    jurisdictions = ['Davis', 'Winters', 'Woodland', 'Yolo County']
    
    for i, year in enumerate(range(2019, 2025)):
        vmt_sheet = f'VMT_{year}'
        per_capita_sheet = f'VMT_per_capita_{year}'
        df_vmt = data_dict[vmt_sheet]
        df_per_capita = data_dict[per_capita_sheet]
        
        # ids.append(f"{year_id}-Total")
        # parents.append(year_id)
        year_id = f"{year}"
        ids.append(year_id)
        parents.append('Total')
        yearly_total = df_vmt[['January', 'February', 'March', 'April', 'May', 'June', 
                              'July', 'August', 'September', 'October', 'November', 'December']].sum().sum()
        values.append(yearly_total)
        labels.append(f'Year {year}<br>VMT: {yearly_total:,.0f}')
        sort_keys.append(i+1)

        for j, jurisdiction in enumerate(jurisdictions):
            row_vmt = df_vmt[df_vmt['Jurisdiction'] == jurisdiction].iloc[0]
            row_per_capita = df_per_capita[df_per_capita['Jurisdiction'] == jurisdiction].iloc[0]
            
            jurisdiction_id = f"{year}-{jurisdiction}"
            jurisdiction_total = row_vmt[['January', 'February', 'March', 'April', 'May', 'June', 
                                        'July', 'August', 'September', 'October', 'November', 'December']].sum()
            per_capita_avg = row_per_capita[['January', 'February', 'March', 'April', 'May', 'June', 
                                           'July', 'August', 'September', 'October', 'November', 'December']].mean()
            
            ids.append(jurisdiction_id)
            parents.append(year_id)
            values.append(jurisdiction_total)
            labels.append(f"{jurisdiction}<br>VMT: {jurisdiction_total:,.0f}<br>Per Capita: {per_capita_avg:.1f}")
            sort_keys.append((i+1)*100 + j)

    df_sunburst = pd.DataFrame({
        'id': ids,
        'parent': parents,
        'value': values,
        'label': labels,
        'sort': sort_keys
    }).sort_values('sort')

    fig = px.sunburst(
        df_sunburst,
        ids='id',
        parents='parent',
        values='value',
        names='label',
        title="Yolo County VMT Distribution (2019-2024)",
    )

    fig.update_layout(
        width=1000, 
        height=1000,
        uniformtext=dict(minsize=10, mode='hide'),
        font=dict(size=12)
    )

    output_path = os.path.join(output_folder, "vmt_sunburst.html")
    # fig.write_html(output_path)
    print(f"Sunburst chart saved to: {output_path}")
    return df_sunburst, fig

base_folder = r"I:\Projects\Warren\Replica Dashboard"
excel_file = "Yolo_Resident_VMT_19_24.xlsx"
filepath = os.path.join(base_folder, excel_file)

data = read_vmt_data(filepath)
df, sunburst = create_sunburst(data, base_folder)

```
