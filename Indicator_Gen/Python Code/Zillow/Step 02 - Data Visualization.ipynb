```python
# Set indicator
indicator_name = 'Cost_2'

file_name = f"{indicator_name} MPO Zillow.xlsx"
df_mpo = pd.read_excel(os.path.join(path_plots, 'Data', file_name), sheet_name = 'Annual')

display(df_mpo.head())
```

```python
color_map = {
    'Sacramento': '#9DC209'
    , 'Yuba City': '#1E90FF'
    , 'Los Angeles': '#FBB117'
    , 'Riverside': '#DC381F'
    , 'San Diego': '#00BFFF'
    , 'San Francisco': '#3B9C9C'
    , 'San Jose': '#B666D2'
}
```

```python
def re_remove_post(x, exp = ','):
    if x == 'nan':
        return 'nan'
    else:
        return x.split(exp, 1)[0]
```

```python
df_plot = df_mpo.copy()


df_plot['Region'] = df_plot['Region'].apply(re_remove_post)
df_plot = df_plot[~df_plot['Region'].str.contains('Median')]
df_plot = df_plot[df_plot['Region'].isin(list(color_map.keys()))]


df_plot['Region_sort'] = pd.Categorical(df_plot['Region'], [
    'Sacramento'
    , 'Yuba City'
    , 'Los Angeles'
    , 'Riverside'
    , 'San Diego'
    , 'San Francisco'
    , 'San Jose'
])

df_plot = df_plot.sort_values(['Region_sort', 'Year'], ascending = [True, False])
df_plot = df_plot.drop(['Region_sort'], axis = 1)
df_plot = df_plot.dropna()
df_plot = df_plot.reset_index(drop = True)
df_plot['Price'] = round(df_plot['Price'])

df_plot['Price'] = df_plot['Price'].astype(int)


display(df_plot.head())

fig = px.line(df_plot, x = 'Year', y = 'Price', color = 'Region', markers = True, color_discrete_map=color_map)

title = "<b>Average Monthly Rents by CA MSA's</b>" 
fig.update_layout(legend_title=None, title=title, template=template, font_family=font_family)
fig.update_layout(xaxis_title = None, yaxis_title = None, yaxis=dict(tickfont = dict(size=12)), xaxis=dict(tickfont = dict(size=12)))
fig.update_xaxes(tick0=0, dtick=1)
fig.update_yaxes(tick0=0, dtick=500, tickprefix='$', tickformat = ',.0f')
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend=dict(orientation="h", yanchor="bottom", y=-0.2,xanchor="right", x=0.9))

fig.show()

# fig.write_html(os.path.join(path_plots, 'Cost_2_rent_price.html'))
```
