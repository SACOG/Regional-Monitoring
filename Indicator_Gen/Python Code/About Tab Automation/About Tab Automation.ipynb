```python
import pandas as pd
import os
import yaml
import datetime
```

```python
# Define user
user = os.getlogin()
path_users = os.path.join('C:\\Users', user)

# Set file paths

# SharePoint
path_sp   = os.path.join(path_users, 'Sacramento Area Council of Governments\Regional Monitoring and Reporting - Documents', 'Data', 'About Indicators.xlsx')
```

```python
excel_file = pd.ExcelFile(path_sp)

# Initialize the main dictionary
dict_about = {}

# Loop through each sheet, skipping the first one
for sheet_name in excel_file.sheet_names[1:]:
    # Read the sheet into a DataFrame without skipping rows
    df = pd.read_excel(excel_file, sheet_name=sheet_name, header = None)

    # Initialize a dictionary to hold information
    info_dict = {}
    notes = []  # To collect notes that span multiple rows

    # Process each row in the DataFrame
    for index, row in df.iterrows():
        key = row.iloc[0]  # Use iloc[0] to correctly access the first column by position
        value = row.iloc[1]  # Use iloc[1] to correctly access the second column by position

        # Check if the row is a note (if key is NaN and value is present)
        if pd.isna(key) and not pd.isna(value):
            notes.append(value)  # Collect the notes into a list
        else:
            # Ensure values are lists for consistency
            if not isinstance(value, list):
                value = [value]

            # Include the 'Indicator Title' and other key rows specifically
            if pd.notna(key) and key == 'Indicator Title':
                info_dict[key] = value  # Ensure the 'Indicator Title' is captured

            # Add collected notes to the dictionary under the 'Notes' key
            if notes:
                info_dict['Notes'] = ' '.join(notes)  # Combine and add notes as a single string
                notes = []  # Reset notes after adding

            # Add the current key-value pair to the dictionary if the key is valid
            if pd.notna(key):
                info_dict[key] = value

    # Add any remaining notes to the dictionary
    if notes:
        info_dict['Notes'] = ' '.join(notes)

    # Use the sheet name directly as the key in the dictionary
    dict_about[sheet_name] = info_dict

```

```python
# Convert the dictionary to a YAML string
yaml_string = yaml.dump(dict_about, sort_keys=False, default_flow_style=False)

# Print the YAML string (optional)
print(yaml_string)

# Save the YAML string to a file
with open('dict_about.yaml', 'w') as yaml_file:
    yaml.dump(dict_about, yaml_file, sort_keys=False, default_flow_style=False)
```
