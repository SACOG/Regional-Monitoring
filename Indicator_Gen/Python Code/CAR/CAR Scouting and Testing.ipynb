# **CAR Data Scouting**

***

Looking into how usable CAR data is. Did not find anything on the existence of a CAR API, but there are manually downloadable files available online. These files are given by link to a SharePoint Excel file. From there, users can download using browser. The data is in an easily readable format, and would require very minimal cleaning. All data is updated with the last month's data by the third week of the current month. Release schedule is [here](https://carorg.sharepoint.com/:b:/s/CAR-RE-PublicProducts/EQHS3YCtUGFEkM-P4MAs-GgBsbl2PSfPVuOugdtWi1nUXA?e=l5bRe0). 

## [Historical Data](https://www.car.org/marketdata/data/housingdata)

- [Sale of Existing Single Family Homes](https://carorg.sharepoint.com/:x:/s/CAR-RE-PublicProducts/EcA8aeskXYNFtkg1Q2Vtl5QBGCIYEGsnv9DsjR3_tGJ0jQ?e=cgxngp).
This data includes the Percent Change of Sales of Existing Detached Homes across January 1990 to August 2024. The data in the excel does not show for 1990, but does when read in. This data includes every county within California. There are some counties that do not have data, but for the most part it is well documented. 

- [Median Prices of Existing Single Family Homes](https://carorg.sharepoint.com/:x:/s/CAR-RE-PublicProducts/EQr6lTtx5_hCmRawI-XHBkoBt0otrc47S8XAem4f8UAMrw?e=thXabK). This dataset includes the Median Prices of Existing Detached Homes across January 1990 to August 2024. Should be noted that Yuba, Sutter, and Yolo have a lot of missing data for certain years. 

- [Unsold Inventory Index (UII) of Existing Single Family Homes](https://carorg.sharepoint.com/:x:/s/CAR-RE-PublicProducts/EcDFUV4JrPtFofTmqhdiPCoBQ7S1MVK9IFU3A6BE6mt9hA?e=SILk7K). This dataset includes the Unsold Inventory Index of Existing Detached Homes from January 1990 to August 2024.

- [Median Time on Market of Existing Single Family Homes](https://carorg.sharepoint.com/:x:/s/CAR-RE-PublicProducts/EaAOiWPma41ArHBWwYRGQMoBlUe5JMxKixyQ0idvrn7i7Q?e=zu8DEk). This dataset includes the Median Time on Market of Existing Detached Homes from January 1990 to August 2024. The data appears to be in terms of days. Should be noted that the SACOG counties have a good chunk of missing data. Additionally, certain metropolitans like the Bay and LA are clumped together. 

## [City Level Sales and Price Data](https://www.car.org/marketdata/interactive/housingmarketoverview)

This data is sourced by using the interactive, PowerBI dashboard in the browser. The instructions taken from the browser are given here. 

1. Using the drop down filter, find and select the county you are interested in. The tables will then display the current reports available for that county and its cities.

2. Next, click on the icon of your preferred format. 

3. Download and save your report. 

For our sake, we probably only need to use the Historical Data tools, as the report is something we would want to do on our own.

```python
# Let's join all of these dataframes together

full_merged = percent_long.merge(index_long, on=['Mon-Yr', 'County', 'MPO']).merge(market_long, on=['Mon-Yr', 'County', 'MPO']).merge(median_long, on=['Mon-Yr', 'County', 'MPO'])

# sacog_merged = sacog_percent_long.merge(sacog_index_long, on=['Mon-Yr', 'County']).merge(sacog_market_long, on=['Mon-Yr', 'County']).merge(sacog_median_long, on=['Mon-Yr', 'County'])

# # Plot all counties
# fig = px.line(sacog_merged, x='Mon-Yr', y='Percent Change', color='County', facet_col='County', template='plotly_white',
#                facet_col_wrap=2, title='Percent Change of Sales for Existing Detached Homes by SACOG County',
#                  labels={
#                      "Percent Change": "Percent Change"
#                  })

# fig.update_layout(yaxis_title = None)

# fig.show()
```

```python
# Need to average each county by combining the median prices for the years

weight_meds = full_merged
weight_meds['Year'] = weight_meds['Mon-Yr'].dt.year

weight_meds = weight_meds.groupby(['Year', 'County', 'MPO']).agg(
    median_price=('Median Price', 'median')
).reset_index()
```

```python
# Merging on year, county, and MPO
# Weighing housing cost for future calcs
# Calc total weighted cost and total population per group
# Calc the population weighted average housing cost for each grouping of (Year, County, MPO)

df_weight = pd.merge(weight_meds, pop_5, on=['County', 'Year', 'MPO'])

df_weight['Weighted Price'] = df_weight['median_price'] * df_weight['Population']

df_grouped = df_weight.groupby(['County', 'Year', 'MPO']).apply(
    lambda x: pd.Series({
        'Weighted Avg Price': x['Weighted Price'].sum() / x['Population'].sum()
    })
).reset_index()

display(df_grouped[df_grouped['MPO']=='SACOG'])
```

```python
# Weighting the MPOs by county and 

merged_df = pd.merge(weight_meds, pop_5, on=['County', 'Year', 'MPO'])

merged_df['Weighted Price'] = merged_df['median_price'] * merged_df['Population']

weighted_avg_county_df = merged_df.groupby(['County', 'Year', 'MPO']).apply(
    lambda x: pd.Series({
        'Weighted Avg Price': x['Weighted Price'].sum() / x['Population'].sum()
    })
).reset_index()

display(weighted_avg_county_df)
```

```python
# aggregating these so that we can graph the MPOs

# needs to be float
full_merged['Market Time'] = pd.to_numeric(full_merged['Market Time'], errors='coerce')

mpo_frame = full_merged.groupby(['MPO', 'Mon-Yr'], as_index=False).agg({
    'Percent Change': 'mean',
    'Index': 'mean',
    'Market Time': 'mean',
    'Median Price': 'median' # not going to actually plot this one tho, use other frame for it
})
```

```python
# Median Time on Market plot

fig = px.line(mpo_frame, 
              x='Mon-Yr', 
              y='Market Time', 
              color='MPO', 
              facet_col='MPO', 
              template='plotly_white',
              facet_col_wrap=2, title='Median Time on Market of Existing Detached Homes by MPOs',
              labels={"Market Time": "Time in Days"})

fig.update_layout(
    yaxis_title=None,
    showlegend=True            
)

fig.for_each_annotation(lambda a: a.update(text=''))
fig.for_each_yaxis(lambda y: y.update(title_text=''))


fig.show()
```
