```python
import arcpy

def convert_table_to_gdb():
    # Allows to be overwritten
    arcpy.env.overwriteOutput = True

    # CSV path and Geodatabase path
    csv_layer = "C:\\Users\\patwood\\OneDrive - Sacramento Area Council of Governments\\Task 8. Reproduce Progress Report indicators\\Indicator Data\\Pop_2V2.csv"
    geodatabase = "I:\\Monitoring and Reporting\\PublishTables\\Default.gdb"

    # Converison
    arcpy.conversion.TableToGeodatabase(Input_Table=[csv_layer], Output_Geodatabase=geodatabase)

    print("Table has been successfully converted to Geodatabase format.")

if __name__ == '__main__':
    convert_table_to_gdb()

print('Done ^_^..V..')
```
