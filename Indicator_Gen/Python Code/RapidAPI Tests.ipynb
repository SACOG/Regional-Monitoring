```python
url = "https://ev-charge-finder.p.rapidapi.com/search-by-location"

headers = {
	"x-rapidapi-key": "27e65b7c6fmsh305b51f3aade8c3p1fab15jsn7eb2f3787331",
	"x-rapidapi-host": "ev-charge-finder.p.rapidapi.com"
}

city_list = ['Sacramento', 'Yuba City', 'Sutter', 'Placerville', 'Davis']
dfs = []
for city in city_list:
    querystring = {"near": city + ', CA, USA', "limit": "150"}
    response = requests.get(url, headers=headers, params=querystring)

    # Flatten the dictionary
    
    flattened_data = []
    for entry in response.json()['data']:
        flattened_entry = {}
        for key, value in entry.items():
            if isinstance(value, dict):
                for sub_key, sub_value in value.items():
                    flattened_entry[f"{key}_{sub_key}"] = sub_value
            else:
                flattened_entry[key] = value
        flattened_data.append(flattened_entry)

    # Convert to df and append to list

    df = pd.DataFrame(flattened_data)
    dfs.append(df)

# Combine all DataFrames into one

combined_df = pd.concat(dfs, ignore_index=True)

# Display the combined DataFrame

print(combined_df.head())
```

```python
combined_df
```

```python
# Function to get latitude and longitude

# Needs to use GeoPy library to convert Lat Longs

geolocator = Nominatim(user_agent="Geopy Library")

def get_lat_lon(address):
    try:
        location = geolocator.geocode(address, timeout=10)
        if location:
            return location.latitude, location.longitude
        else:
            return None, None
    except GeocoderTimedOut:
        return get_lat_lon(address)  # Retry if geocoding times out

# Apply the function
combined_df['Latitude'], combined_df['Longitude'] = zip(*combined_df['formatted_address'].apply(get_lat_lon))
```

```python
combined_df = combined_df.dropna(subset=['Latitude', 'Longitude'])
```

```python
map_center = [38.5768267, -121.488201]  # SACOG Address
mymap = folium.Map(location=map_center, zoom_start=12)

# Adding the markers on the map

for _, row in combined_df.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=f"<strong>{row['name']}</strong><br>"
              f"Address: {row['formatted_address']}",
        tooltip=row['name']
    ).add_to(mymap)

# Save the map, lowk don't need to do tho

mymap.save("charging_stations_map.html")
```
