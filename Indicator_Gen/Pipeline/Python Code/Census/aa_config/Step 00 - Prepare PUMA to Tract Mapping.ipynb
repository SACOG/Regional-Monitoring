```python
# General
import numpy as np
import pandas as pd
import os
import requests
```

```python
# Working directory
rootpath = os.path.dirname(os.getcwd())
print(rootpath)

# Set file paths
path_out = os.path.join(rootpath, 'aa_config')
```

```python
# Use URL to county fips mapping table
# Import county FIPS codes by state
url_puma = "https://www2.census.gov/geo/docs/maps-data/data/rel2020/2020_Census_Tract_to_2020_PUMA.txt"
df_puma = pd.read_csv(url_puma, header = 0, sep = ',')
df_puma.head()
```

```python
# reformat FIPS fields

df_puma['PUMA5CE' ] = df_puma['PUMA5CE' ].astype(str).apply('{:0>5}'.format)
df_puma['TRACTCE' ] = df_puma['TRACTCE' ].astype(str).apply('{:0>6}'.format)
df_puma['COUNTYFP'] = df_puma['COUNTYFP'].astype(str).apply('{:0>3}'.format)
df_puma['STATEFP' ] = df_puma['STATEFP' ].astype(str).apply('{:0>2}'.format)

# show
df_puma.head()
```

```python
# Export locally just in case url gets moved
df_puma.to_excel(os.path.join(path_out, 'PUMA to Tract FIPS Code Mapping.xlsx'), index=False)
```
