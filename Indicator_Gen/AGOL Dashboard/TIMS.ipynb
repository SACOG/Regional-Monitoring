```python

indicator = 'Safety_1'
plot_name = 'fatality_rate'

df_mpo   = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} MPO TIMS.xlsx"      , sheet_name = 'MPO'      )
df_state = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} Statewide TIMS.xlsx", sheet_name = 'Statewide')

df_mpo = df_mpo[df_mpo['MPO']=='SACOG']
df_mpo  ['Geography'] = 'Sacramento Region'
df_state['Geography'] = 'California'

df_plot = (
            pd.concat([df_mpo, df_state])
                .drop(['MPO', 'State'], axis=1)
                .sort_values(['Geography', 'Year'], ascending=[False, True])
                .reset_index(drop=True)
        )


fig = go.Figure()

# Plot traces
for geography, color in pt.colors_ca_tims.items():
    df_cat = df_plot[df_plot['Geography'] == geography].sort_values('Year', ascending=True)
    df_cat = df_cat.dropna(subset=['Fatalities_100 mvmt_5 Year Rolling Average'])

    fig.add_trace(go.Bar(
        x=df_cat['Year'],
        y=df_cat['Fatalities_100 mvmt_5 Year Rolling Average'],
        name=geography,
        marker=dict(color=color),
        hovertemplate='%{y:.1f}'
    ))

for geography, color in pt.colors_ca_tims.items():
    if geography == 'California':
        continue
    df_cat = df_plot[df_plot['Geography'] == geography].sort_values('Year', ascending=True)
    df_cat = df_cat.dropna(subset=['Fatalities_5 Year Rolling Average'])
    
    fig.add_trace(go.Bar(
        x=df_cat['Year'],
        y=df_cat['Fatalities_5 Year Rolling Average'],
        name=geography,
        marker=dict(color=color),
        hovertemplate='%{y:,.0f}',
        visible=False
    ))

# Create buttons for toggling: also update y-axis tickformat
buttons = [
    dict(
        label='Rate Per 100 MVMT',
        method='update',
        args=[
            {'visible': [True, True, False]},
            {'yaxis': {'tickformat': '.1f', 'range': [0, 2.1], 'dtick': .5},
             'annotations': [dict(ann, visible=True)]}
        ]
    ),
    dict(
        label='Total',
        method='update',
        args=[
            {'visible': [False, False, True]},
            {'yaxis': {'tickformat': ',.0f', 'range': [0, 410], 'dtick': 50},
             'annotations': [dict(ann, visible=True)]}
        ]
    )
]

# Add dropdown menu to layout
fig.update_layout(
    annotations=[dict(ann, visible=True)],
    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            pad={"r": 10, "t": 10},
            showactive=True,
            x=1.25,
            xanchor="right",
            y=1.275,
            yanchor="top"
        )
    ],
    # Initial axis formatting should match the initially visible set
    xaxis=dict(dtick=1),
    yaxis=dict(tickformat='.1f', range=[0,2.1], dtick=0.5),
    title_text='Traffic Fatalities, Five Year Rolling Average',
    legend=dict(orientation="h", yanchor="bottom", y=-0.2, xanchor="right", x=0.5)
)

title='<b>Traffic Fatalities, Five Year Rolling Average</b> <br><sup>Six-County Sacramento Region</sup>'
pt.plot_agol(fig, EXPORT, title, indicator, plot_name)

```
