```python

indicator = 'Safety_1'
plot_name = 'toggle_fatality_rate'


df_mpo   = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} MPO TIMS.xlsx"      , sheet_name = 'MPO'      )
df_state = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} Statewide TIMS.xlsx", sheet_name = 'Statewide')


df_plot1 = df_mpo  .copy()
df_plot2 = df_state.copy()


df_plot1 = df_plot1.rename(columns = {'MPO'  :'Geography'})
df_plot1['Geography'] = 'Sacramento Region'
df_plot2 = df_plot2.rename(columns = {'State':'Geography'})
df_plot2['Geography'] = 'California'

df_plot = pd.concat([df_plot1, df_plot2])
df_plot = df_plot.sort_values(['Geography', 'Year'], ascending = [False, True])
df_plot = df_plot.reset_index(drop = True)

display(df_plot.head(), df_plot.tail())


color_map  = {
    'Sacramento Region': '#9DC209'
    , 'California': '#1E90FF'
}

fig=go.Figure()

for geography, color in color_map.items():
    df_cat = df_plot[df_plot['Geography'] == geography]
    df_cat=df_cat.sort_values('Year',ascending=True)
    df_cat=df_cat.dropna(subset=['Fatalities 100 mvmt_5 Year Rolling Average'])

    fig.add_trace(go.Bar(
        x=df_cat['Year'],
        y=df_cat['Fatalities 100 mvmt_5 Year Rolling Average'],
        name=geography,
        marker=dict(color=color),
        hovertemplate='%{y}'
    ))

for geography, color in color_map.items():
    df_cat = df_plot[df_plot['Geography'] == geography]
    df_cat=df_cat.sort_values('Year',ascending=True)
    df_cat=df_cat.dropna(subset=['Fatalities 100 mvmt'])
    
    fig.add_trace(go.Bar(
        x=df_cat['Year'],
        y=df_cat['Fatalities 100 mvmt'],
        name=geography,
        marker=dict(color=color),
        hovertemplate='%{y}',
        visible=False
    ))

# Create buttons for toggling
buttons = [
    dict(
        label='5 Year Average',
        method='update',
        args=[
            {'visible': [True, True,False, False]},
            {'xaxis': {'tickangle': 0}}
        ]
    ),
    dict(
        label='Annual',
        method='update',
        args=[
            {'visible': [False, False, True, True]},
            {'xaxis': {'tickangle': 0}}
        ]
    )
]

# Add dropdown menu to layout
fig.update_layout(
    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            pad={"r": 10, "t": 10},
            showactive=True,
            x=1,
            xanchor="right",
            y=1.1,
            yanchor="top"
        )
    ]
)

title = 'Traffic Fatality Rates'
fig.update_yaxes(tick0=0, dtick=0.2, range = [0.5, 2.1])
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend=dict(orientation="h", yanchor="bottom", y=-0.15,xanchor="right", x=0.55))

pt.plot_agol(fig, export, title, indicator, plot_name)
```

```python
export = False
indicator_name = 'Safety_3'
plot_name = 'ej'


df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator_name} TIMS.xlsx", sheet_name='Region', skiprows=1)
df_mpo = df_mpo[['Year', 'Fatal', 'Serious Injury', 'Total Injury']]


df_plot = df_mpo.copy()

df_plot.columns = ['Year', 'Fatal Collisions', 'Serious Injury Collisions', 'Total Injury Collisions']
df_plot = pd.melt(df_plot, id_vars = ['Year'], var_name = 'Metric', value_name = 'Percentage')
df_plot['Percentage'] = round(df_plot['Percentage']*100, 1)

display(df_plot.head())


color_map = {
    'Fatal Collisions': '#1F45FC'
    , 'Serious Injury Collisions': '#1E90FF'
    , 'Total Injury Collisions': '#9DC209'
}

fig = px.bar(df_plot, y='Percentage', x='Year'
             , color='Metric'
             , facet_col='Metric'
             , color_discrete_map=color_map)

title = '<b>Percent of Collisions Occurring in Environmental Justice Communities</b>  <br><sup>Non-freeway collisions in 6-county Sacramento region</sup><br><br>'
fig.update_yaxes(tick0=0, dtick=10, range = [0, 102], ticksuffix='%')
fig.update_traces(hovertemplate='%{y}')
fig.update_xaxes(dtick=2)
fig['layout']['xaxis2']['title']['text']=''
fig['layout']['xaxis3']['title']['text']=''
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))

for annotation in fig['layout']['annotations']:
    annotation['y'] = -0.1
    
pt.plot_agol(fig, export, title, indicator, plot_name)
```

```python
export = True
import plot as pt
indicator = 'Safety_4'
plot_name = 'toggle_fatalities'


df_mpo = pd.read_excel(os.path.join(PATH_PLOTS/'Data'/f"{indicator} MPO TIMS.xlsx"), sheet_name='MPO')



df_plot = df_mpo.copy()

df_plot = df_plot.dropna()
df_plot = df_plot.sort_values(['Year', 'Category'], ascending = [False, True])

display(df_plot.head())


color_map  = {
    'Bicyclists': '#9DC209'
    , 'Pedestrians': '#1E90FF'
}

fig=go.Figure()

for category, color in color_map.items():
    df_cat = df_plot[df_plot['Category'] == category]
    df_cat=df_cat.sort_values('Year',ascending=True)
    fig.add_trace(go.Bar(
        x=df_cat['Year'],
        y=df_cat['Fatalities_5 Year Average'],
        name=category,
        marker=dict(color=color),
        hovertemplate='%{y}'
    ))

for category, color in color_map.items():
    df_cat = df_plot[df_plot['Category'] == category]
    df_cat=df_cat.sort_values('Year',ascending=True)
    df_cat=df_cat[df_cat['Fatalities']!=0]
    
    fig.add_trace(go.Bar(
        x=df_cat['Year'],
        y=df_cat['Fatalities'],
        name=category,
        marker=dict(color=color),
        hovertemplate='%{y}',
        visible=False
    ))

# Create buttons for toggling
buttons = [
    dict(
        label='5 Year Average',
        method='update',
        args=[
            {'visible': [True, True,False, False]},
            {'xaxis': {'tickangle': 0}}
        ]
    ),
    dict(
        label='Annual',
        method='update',
        args=[
            {'visible': [False, False, True, True]},
            {'xaxis': {'tickangle': 0}}
        ]
    )
]

# Add dropdown menu to layout
fig.update_layout(
    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            pad={"r": 10, "t": 10},
            showactive=True,
            x=1,
            xanchor="right",
            y=1.1,
            yanchor="top"
        )
    ]
)


title = 'Fatalities'
fig.update_yaxes(tick0=0, dtick=10, range = [0, 101])
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend=dict(orientation="h", yanchor="bottom", y=-0.15,xanchor="right", x=0.55))

pt.plot_agol(fig, export, title, indicator, plot_name)

```
