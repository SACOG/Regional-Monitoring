```python
# Set indicator
indicator_name = 'Safety_3'
plot_name = 'ej'
export = False


## Importing ---

file_name = f"{indicator_name} TIMS.xlsx"
df_mpo = pd.read_excel(os.path.join(path_plots, 'Data', file_name), sheet_name = 'Region', skiprows = 1)
df_mpo = df_mpo[['Year', 'Fatal', 'Serious Injury', 'Total Injury']]


## Organizing ---

df_plot = df_mpo.copy()

df_plot.columns = ['Year', 'Fatal Collisions', 'Serious Injury Collisions', 'Total Injury Collisions']
df_plot = pd.melt(df_plot, id_vars = ['Year'], var_name = 'Metric', value_name = 'Percentage')
df_plot['Percentage'] = round(df_plot['Percentage']*100, 1)

display(df_plot.head())


## Plotting ---

color_map = {
    'Fatal Collisions': '#1F45FC'
    , 'Serious Injury Collisions': '#1E90FF'
    , 'Total Injury Collisions': '#9DC209'
}

fig = px.bar(df_plot, y='Percentage', x='Year'
             , color='Metric'
             , facet_col='Metric'
             , color_discrete_map=color_map)

title = '<b>Percent of Collisions Occurring in Environmental Justice Communities, 2020</b>  <br><sup>Non-freeway collisions in 6-county Sacramento region</sup><br><br>'
fig.update_yaxes(tick0=0, dtick=10, range = [0, 102], ticksuffix='%')
fig.update_traces(hovertemplate='%{y}')
fig.update_xaxes(dtick=2)
fig['layout']['xaxis2']['title']['text']=''
fig['layout']['xaxis3']['title']['text']=''
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))

for annotation in fig['layout']['annotations']:
    annotation['y'] = -0.1
    
plot_agol(export=export)
```
