```python


import pandas as pd
from pathlib import Path
import plotly.express as px
from IPython.display import display
import sys
sys.path.append(str(Path.home() / 'Documents' / 'Projects' / 'Regional-Monitoring' / 'Indicator_Gen' / 'config'))
import plot as pt

PATH_PLOTS = Path(r"\\webmapping-svr\c$\inetpub\wwwroot\monitoring")
print('EXPORT Location: ' + str(PATH_PLOTS))


def clean_fdic(df):
    
    df = df[df['Banking Type']!='No Full Access (%)']
    df['Banking Type'] = df['Banking Type'].str.replace(' (%)', '')
    df['Percent'] = df['Percent']*100

    return df


EXPORT=False
```

```python

indicator = 'Income_8'
plot_name = 'MSA'
geography = 'MSA'

df = pd.read_excel(PATH_PLOTS / 'Data' / f"{indicator} FDIC.xlsx", sheet_name=geography)
df = df.melt(id_vars=['Year'], var_name='Banking Type', value_name='Percent')
df = clean_fdic(df)
df = df.sort_values(['Year', 'Percent'], ascending=[True, False])
df['Year'] = df['Year'].astype(str)
display(df.head())

fig = px.bar(df, x='Year', y='Percent', color='Banking Type', color_discrete_map=pt.colors_banking)
title = '<b>Households Without Full Banking Access</b>  <br><sup>Four-County Sacramento MSA</sup>'
fig.update_yaxes(tick0=0, dtick=5, range=[0,36], ticksuffix='%')
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```
