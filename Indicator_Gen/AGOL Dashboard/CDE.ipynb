```python
# Set Indicator
indicator_name = 'Edu_3'
plot_name = 'A_G'
export = False


## Importing ---

file_name = f"{indicator_name} A-G.xlsx"
df_mpo = pd.read_excel(os.path.join(path_plots, 'Data', file_name), sheet_name = 'Indicator', skiprows=1, usecols="AB:AI")


## Organizing ---
df_plot = df_mpo.copy()


df_plot.columns = ['Race_Ethnicity', '2016-2017', '2017-2018', '2018-2019', '2019-2020', '2020-2021', '2021-2022', '2022-2023']
df_plot = pd.melt(df_plot, id_vars = 'Race_Ethnicity', var_name = 'Year', value_name = 'Percentage')

df_plot['Sort1'] = pd.Categorical(df_plot['Race_Ethnicity'], [
    'Asian'
    , "Black"
    , "Latino"
    , 'White'
    , 'Socioeconomically Disadvantaged'
    , 'Total'
])

df_plot['Sort2'] = pd.Categorical(df_plot['Year'], [
   '2016-2017', '2017-2018', '2018-2019', '2019-2020', '2020-2021', '2021-2022', '2022-2023'
])
    
df_plot = df_plot.sort_values(['Sort2', 'Sort1'], ascending = [True, True])
df_plot = df_plot.drop(['Sort2', 'Sort1'], axis = 1)
df_plot['Percentage'] = round(df_plot['Percentage']*100, 1)
df_plot = df_plot.reset_index(drop = True)

display(df_plot.head())



## Plotting ---
color_map  = {
    'Asian': '#9DC209'
    , 'Black': '#1E90FF'
    , 'Latino': "#FBB117"
    , 'White': "#DC381F"
    , 'Socioeconomically Disadvantaged': '#7E587E'
    , 'Total': '#1F45FC'
}


fig = px.bar(df_plot, y='Percentage', x='Year'
             , color='Race_Ethnicity'
             , color_discrete_map=color_map
             , barmode = 'group')


title = '<b>Percent Meeting A-G Requirements, by Race/Ethnicity and Socioeconomic Status</b>  <br><sup>6-County Sacramento Region</sup> '
fig.update_yaxes(tick0=0, dtick=10, ticksuffix = '%', range=[0, 101])
fig.update_xaxes(tick0=0, dtick=1)
fig.update_traces(hovertemplate='%{y}')


plot_agol(export=export)
```
