```python

indicator = 'Congestion_1'
plot_name = 'PHED_UZA'
year=2024


df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} UZA RITIS.xlsx", sheet_name='UZA')

df_plot = df_mpo.copy()
df_plot['Month'] = pd.to_datetime(df_plot['Month']) #format = "%Y/%m"
df_plot['Month'] = df_plot['Month'].dt.to_period('M')
df_plot['Year'] = df_plot['Month'].dt.year
df_plot = df_plot[df_plot['Year'] != 2011]
df_plot = df_plot.groupby(['Year', 'UZA']).sum(numeric_only=True).reset_index()
df_plot = df_plot[['Year', 'UZA', 'PHED (hours)']]
df_plot = df_plot[df_plot['Year'] == year].sort_values('PHED (hours)', ascending=False).reset_index(drop=True)
display(df_plot.head())

fig = px.bar(df_plot, y='UZA', x='PHED (hours)', color='UZA', color_discrete_map=pt.colors_peers, orientation='h')
ticktext = []
for geography in df_plot['UZA'].unique():
    if geography in ['Sacramento']:
        ticktext.append(f'<b>{geography}</b>')
    else:
        ticktext.append(geography)
fig.update_layout(yaxis=dict(tickmode='array', tickvals=df_plot['UZA'].unique(), ticktext=ticktext))
title = f'<b>Annual Per Capita Person Hours of Excessive Delay, {year}</b>  <br><sup>Peer Regions at UZA Geography</sup>'
fig.update_xaxes(tick0=0, dtick=2)
fig.update_traces(hovertemplate='%{x}')
fig.update_layout(showlegend=False)


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```

```python


indicator = 'Congestion_1'
plot_name = 'PHED'

df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} UZA RITIS.xlsx", sheet_name='UZA')

df_plot = df_mpo.copy()
df_plot['Month'] = pd.to_datetime(df_plot['Month']) #format = "%Y/%m"
df_plot['Month'] = df_plot['Month'].dt.to_period('M')
df_plot['Year'] = df_plot['Month'].dt.year
df_plot = df_plot[df_plot['Year'] != 2011]
df_plot = df_plot.groupby(['Year', 'UZA']).sum(numeric_only=True).reset_index()
df_plot = df_plot[['Year', 'UZA', 'PHED (hours)']]
df_plot = df_plot[
    (df_plot['Year'] >= 2016) 
    & (df_plot['Year'] <  2025)
    & (df_plot['UZA'] == 'Sacramento')
    ].reset_index(drop=True)

display(df_plot.head())


fig = px.bar(df_plot, y='PHED (hours)', x='Year')
fig.update_traces(marker_color='#27AAE1')
title = '<b>Annual Person Hours of Delay Per Capita</b>  <br><sup>National Highway System facilities in Sacramento Urbanized Area</sup>'
fig.update_xaxes(tick0=0, dtick=1)
fig.update_yaxes(tick0=0, dtick=5, range=[0, 21])
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(showlegend=False)


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```

```python


indicator = 'Congestion_3'
plot_name = 'LOTTR'

df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} SACOG RITIS.xlsx", sheet_name='SACOG')

df_plot = df_mpo.copy()
df_plot['Month'] = pd.to_datetime(df_plot['Month']) #format = "%Y/%m"
df_plot['Month'] = df_plot['Month'].dt.to_period('M')
df_plot['Year'] = df_plot['Month'].dt.year
df_plot = df_plot[df_plot['Year'] >= 2016]
df_plot = df_plot.groupby(['Year'], as_index = False).mean(numeric_only=True)
df_plot = df_plot[['Year', 'Interstate LOTTR (%)',	'Non-Interstate LOTTR (%)']]
df_plot.columns = ['Year', 'Interstate', 'Non-Interstate']
df_plot = pd.melt(df_plot, id_vars = 'Year')
df_plot['value'] = round(df_plot['value'], 1)
df_plot = df_plot.reset_index(drop=True)

display(df_plot.head())


fig = px.bar(df_plot, y='value', x='Year', color='variable', color_discrete_map=pt.colors_hwy, barmode='group')
title = '<b>Percent of Reliable Roads and Freeways</b>  <br><sup>National Highway System facilities in Sacramento Urbanized Area</sup>'
fig.update_xaxes(tick0=0, dtick=1)
fig.update_yaxes(tick0=0, dtick=20, ticksuffix='%', range = [0, 102])
fig.update_traces(hovertemplate='%{y}')


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```

```python


indicator = 'Congestion_3'
plot_name = 'TTTR'

df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} SACOG RTIS.xlsx", sheet_name='SACOG')


df_plot = df_mpo.copy()
df_plot['Month'] = pd.to_datetime(df_plot['Month']) #format = "%Y/%m"
df_plot['Month'] = df_plot['Month'].dt.to_period('M')
df_plot['Year'] = df_plot['Month'].dt.year
df_plot = df_plot.groupby(['Year'], as_index = False).mean(numeric_only=True)
df_plot = df_plot[df_plot['Year'] >= 2014]
df_plot = df_plot[['Year', 'TTTR (%)']]
df_plot['TTTR (%)'] = round(df_plot['TTTR (%)'], 2)
df_plot = df_plot.reset_index(drop=True)

display(df_plot.head())


fig = px.bar(df_plot, y='TTTR (%)', x='Year')
fig.update_traces(marker_color='#27AAE1')
title = '<b>Truck Time Reliability Index</b>  <br><sup>Sacramento Urbanized Area</sup>'
fig.update_xaxes(tick0=0, dtick=1)
fig.update_yaxes(tick0=0, dtick=0.5, range = [0, 2.2])
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(showlegend=False)


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```
