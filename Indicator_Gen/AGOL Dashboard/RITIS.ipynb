```python
export=True
# Set Indicator
indicator = 'Congestion_2 Monthly'
plot_name = 'percent_road_miles'

## Importing ---
file_name = f"{indicator} SACOG RITIS.xlsx"
df = pd.read_excel(os.path.join(PATH_PLOTS, 'Data', file_name), sheet_name = 'SACOG')

## Organizing ---
df['pct_miles_congested'] = round(df['pct_miles_congested'], 1)
df = df.reset_index(drop = True)

# --- MONTHLY DATA ---
# Filter for rows where year_or_month contains underscore (e.g., '2025_01')
df_monthly = df[df['year_or_month'].astype(str).str.contains('_')].copy()

# --- YEARLY DATA ---
# Filter for rows where year_or_month is just a year (no underscore)
df_yearly = df[~df['year_or_month'].astype(str).str.contains('_')].copy()
df_yearly = df_yearly.sort_values('year_or_month')



# Create formatted month label for display (e.g., 'Jan 2025')
df_monthly['month_str'] = df_monthly['year_or_month'].astype(str).apply(
    lambda x: pd.to_datetime(x, format='%Y_%m').strftime('%b %Y')
)
df_monthly = df_monthly.sort_values('year_or_month')

# 2. Keep only the 12 most recent months
df_monthly = df_monthly.tail(12)

# Create figure with both traces
fig = go.Figure()

# Add yearly trace
fig.add_trace(go.Bar(
    x=df_yearly['year_or_month'].astype(str),
    y=df_yearly['pct_miles_congested'],
    marker_color='#1E90FF',
    name='Yearly Average',
    visible=True,
    hovertemplate='%{y}<extra></extra>'
))

# Add monthly trace
fig.add_trace(go.Bar(
    x=df_monthly['month_str'],
    y=df_monthly['pct_miles_congested'],
    marker_color='#1E90FF',
    name='Monthly',
    visible=False,
    hovertemplate='%{y}<extra></extra>'
))

# Create buttons for toggling
buttons = [
    dict(
        label='Yearly Average',
        method='update',
        args=[
            {'visible': [True, False]},
            {'xaxis': {'tickangle': 0}}
        ]
    ),
    dict(
        label='Monthly',
        method='update',
        args=[
            {'visible': [False, True]},
            {'xaxis': {'tickangle': 0}}
        ]
    )
]

# Add dropdown menu to layout
fig.update_layout(
    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            pad={"r": 10, "t": 10},
            showactive=True,
            x=1,
            xanchor="right",
            y=1.1,
            yanchor="top"
        )
    ]
)

title = '<b>Percent Congested Facilities</b><br><sup>National Highway System facilities in Sacramento Urbanized Area</sup>'
fig.update_yaxes(ticksuffix='%')

pt.plot_agol(fig, export, title, indicator, plot_name)
```

```python

# Set Indicator
indicator_name = 'Congestion_3'
plot_name = 'LOTTR'



## Importing ---

file_name = f"{indicator_name} SACOG RITIS.xlsx"
df_mpo = pd.read_excel(os.path.join(PATH_PLOTS, 'Data', file_name), sheet_name = 'SACOG')


## Organizing ---

df_plot = df_mpo.copy()

df_plot['Month'] = pd.to_datetime(df_plot['Month']) #format = "%Y/%m"
df_plot['Month'] = df_plot['Month'].dt.to_period('M')
df_plot['Year'] = df_plot['Month'].dt.year
df_plot = df_plot[df_plot['Year'] >= 2016]
df_plot = df_plot.groupby(['Year'], as_index = False).mean(numeric_only = True)

df_plot = df_plot[['Year', 'Interstate LOTTR (%)',	'Non-Interstate LOTTR (%)']]
df_plot.columns = ['Year', 'Interstate', 'Non-Interstate']

df_plot = pd.melt(df_plot, id_vars = 'Year')
df_plot['value'] = round(df_plot['value'], 1)

df_plot = df_plot.reset_index(drop = True)

display(df_plot.head())




## Plotting ---

color_map = {
    'Interstate':'#9DC209'
    , 'Non-Interstate':'#1F45FC'
}


fig = px.bar(df_plot, y='value', x='Year', color='variable', color_discrete_map=color_map, barmode='group')


title = '<b>Percent of Reliable Roads and Freeways</b>  <br><sup>National Highway System facilities in Sacramento Urbanized Area</sup>'
fig.update_xaxes(tick0=0, dtick=1)
fig.update_yaxes(tick0=0, dtick=10, ticksuffix='%', range = [0, 102])
fig.update_traces(hovertemplate='%{y}')


pt.plot_agol(fig, export, title, indicator, plot_name)
```
