```python


import pandas as pd
from pathlib import Path
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display
import sys
sys.path.append(str(Path.home() / 'Documents' / 'Projects' / 'Regional-Monitoring' / 'Indicator_Gen' / 'config'))
import plot as pt
import helpers

PATH_PLOTS = Path(r"\\webmapping-svr\c$\inetpub\wwwroot\monitoring")
print('EXPORT Location: ' + str(PATH_PLOTS))
pd.options.display.float_format = '{:.2f}'.format


EXPORT=False
```

```python

indicator = 'VMT_1'
plot_name = 'per_capita_trends'

df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} Per Capita Trends.xlsx", sheet_name='Data', skiprows=8)

df_plot = df_mpo.copy()
df_plot = df_plot.loc[0:4,:]
df_plot = df_plot[['VMT Per Capita'] + helpers.sequence(2006, 2022, 1)]
df_plot = df_plot.rename(columns = {'VMT Per Capita':'MPO'})

df_plot = pd.melt(df_plot, id_vars = 'MPO', var_name = 'Year', value_name = 'VMT Per Capita')
df_plot['VMT Per Capita'] = round(df_plot['VMT Per Capita']*100, 1)
df_plot.loc[df_plot['MPO'] == 'SACOG', 'MPO'] = 'Sacramento'

display(df_plot.head())

fig = px.line(df_plot, x='Year', y='VMT Per Capita', color='MPO', color_discrete_map=pt.colors_ca, markers=True)
title = '<b>Change in Per Capita VMT by CA Region</b>'
fig.update_yaxes(dtick=5, ticksuffix='%', range = [-16,11])
fig.update_xaxes(dtick=1)
fig.update_traces(hovertemplate="%{y}")

pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```

```python

indicator = 'VMT_2'
plot_name = 'average_VMT'

df_plot = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} SACOG Region.xlsx", sheet_name = 'Data')

df_avg = df_plot[['Year', 'Average']]
df_avg['Average'] = round(df_avg['Average'], 1)
df_avg = df_avg[df_avg['Year']>2020]

df_plot = df_plot.drop('Average', axis=1)
df_plot = df_plot.melt(id_vars='Year', var_name='Source', value_name='VMT')

fig = px.line(df_plot, x='Year', y='VMT', color='Source', color_discrete_map=pt.colors_vmt, markers=True)
title = '<b>Daily Vehicle Miles Traveled Per Capita</b>  <br><sup>6-County Sacramento Region</sup>'
fig.add_trace(go.Scatter(x=df_avg['Year'], y=df_avg['Average'], name = 'Average', line=go.scatter.Line(color="#2C3539", dash="dot")))
# fig.update_yaxes(dtick=5, range = [0, 26])
fig.update_xaxes(dtick=1, range = [df_plot['Year'].min()-0.5, df_plot['Year'].max()+0.5])
fig.update_traces(hovertemplate="%{y}")
# fig.update_layout(showlegend=False)


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```
