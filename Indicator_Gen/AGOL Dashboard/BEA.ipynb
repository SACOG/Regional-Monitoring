```python
# Set Indicator
indicator_name = 'Output_1'
plot_name = 'GRP'
export = False



## Importing ---

file_name = f"{indicator_name} MSA BEA.xlsx"
df_msa = pd.read_excel(os.path.join(path_plots, 'Data', file_name), sheet_name = 'Rates')


## Organizing ---

df_plot = df_msa.copy()

df_plot = df_plot[df_plot['MSA'].isin(['Sacramento-Roseville-Folsom'
                                      , 'Yuba City'
                                      , 'Austin-Round Rock-Georgetown'
                                      , 'Charlotte-Concord-Gastonia'
                                      , 'Cincinnati'
                                      , 'Cleveland-Elyria'
                                      , 'Columbus'
                                      , 'Detroit-Warren-Dearborn'
                                      , 'Indianapolis-Carmel-Anderson'
                                      , 'Miami-Fort Lauderdale-Pompano Beach'
                                      , 'Orlando-Kissimmee-Sanford'
                                      , 'Phoenix-Mesa-Chandler'
                                      , 'Portland-Vancouver-Hillsboro'
                                      , 'Riverside-San Bernardino-Ontario'
                                      , 'Salt Lake City'
                                      , 'San Antonio-New Braunfels'
                                      , 'St. Louis'
                                      , 'Tampa-St. Petersburg-Clearwater'])]
df_plot = df_plot.reset_index(drop = True)

df_plot['MSA_sort'] = pd.Categorical(df_plot['MSA'], [
    'Sacramento-Roseville-Folsom'
    , 'Yuba City'
    , 'Austin-Round Rock-Georgetown'
    , 'Charlotte-Concord-Gastonia'
    , 'Cincinnati'
    , 'Cleveland-Elyria'
    , 'Columbus'
    , 'Detroit-Warren-Dearborn'
    , 'Indianapolis-Carmel-Anderson'
    , 'Miami-Fort Lauderdale-Pompano Beach'
    , 'Orlando-Kissimmee-Sanford'
    , 'Phoenix-Mesa-Chandler'
    , 'Portland-Vancouver-Hillsboro'
    , 'Riverside-San Bernardino-Ontario'
    , 'Salt Lake City'
    , 'San Antonio-New Braunfels'
    , 'St. Louis'
    , 'Tampa-St. Petersburg-Clearwater'
])
    
df_plot = df_plot.sort_values(['MSA_sort'], ascending = [True])
df_plot = df_plot.drop(['MSA_sort'], axis = 1)
df_plot['AnnualPctDiff_2017_2022'] = round(df_plot['AnnualPctDiff_2017_2022'], 1)

display(df_plot.head())


## Plotting ---

color_map  = {
    'Sacramento-Roseville-Folsom': '#9DC209'
    , 'Yuba City': '#9DC209'
    , 'Austin-Round Rock-Georgetown': "#1E90FF"
    , 'Charlotte-Concord-Gastonia': "#1E90FF"
    , 'Cincinnati': "#1E90FF"
    , 'Cleveland-Elyria': "#1E90FF"
    , 'Columbus': "#1E90FF"
    , 'Detroit-Warren-Dearborn': "#1E90FF"
    , 'Indianapolis-Carmel-Anderson': "#1E90FF"
    , 'Miami-Fort Lauderdale-Pompano Beach': "#1E90FF"
    , 'Orlando-Kissimmee-Sanford': "#1E90FF"
    , 'Phoenix-Mesa-Chandler': "#1E90FF"
    , 'Portland-Vancouver-Hillsboro': "#1E90FF"
    , 'Riverside-San Bernardino-Ontario': "#1E90FF"
    , 'Salt Lake City': "#1E90FF"
    , 'San Antonio-New Braunfels': "#1E90FF"
    , 'St. Louis': "#1E90FF"
    , 'Tampa-St. Petersburg-Clearwater': "#1E90FF"
}


fig = px.bar(df_plot, y='MSA', x='AnnualPctDiff_2017_2022'
             , color='MSA'
             , color_discrete_map=color_map
             , orientation='h')


ticktext = []
for geography in df_plot['MSA'].unique():
    if geography in ['Sacramento-Roseville-Folsom', 'Yuba City']:
        ticktext.append(f'<b>{geography}</b>')
    else:
        ticktext.append(geography)

fig.update_layout(yaxis=dict(tickmode='array', tickvals=df_plot['MSA'].unique(), ticktext=ticktext))

title = 'Annual Gross Regional Product Growth:  2017-2022'
fig.update_xaxes(tick0=0, dtick=2, ticksuffix='%')
fig.update_traces(hovertemplate='%{x}')
fig.update_layout(showlegend = False)


plot_agol(export=export)
```
