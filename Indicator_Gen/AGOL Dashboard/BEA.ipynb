```python

export=False

import pandas as pd
from pathlib import Path
from IPython.display import display
pd.set_option('display.max_columns', None)

import plotly.express as px
pd.options.display.float_format = '{:.2f}'.format


PATH_GIT = Path.home() / 'Documents' / 'Projects' / 'Regional-Monitoring' / 'Indicator_Gen'
PATH_CONFIG0 = PATH_GIT / 'config'
PATH_PLOTS = Path(r"\\webmapping-svr\c$\inetpub\wwwroot\monitoring")


import sys
sys.path.append(str(PATH_CONFIG0))
import plot as pt


```

```python


indicator = 'Output_1'
plot_name = 'GRP'
year_min = 2001
year_max = 2023
col_year_diff = f'AnnualPctDiff_{year_min}_{year_max}'


df_msa = pd.read_excel(PATH_PLOTS / 'Data' / f"{indicator} MSA BEA.xlsx", sheet_name='Rates')
display(df_msa.head())


df_plot = df_msa.copy()  
df_plot = df_plot.sort_values([col_year_diff], ascending = [False])
df_plot[col_year_diff] = round(df_plot[col_year_diff]*100, 1)
df_plot = df_plot.reset_index(drop=True)

fig = px.bar(df_plot, y='MSA', x=col_year_diff
             , color='MSA'
             , color_discrete_map=pt.color_map_nat_peers
             , orientation='h')


ticktext = []
for geography in df_plot['MSA'].unique():
    if geography in ['Sacramento, CA', 'Yuba City, CA']:
        ticktext.append(f'<b>{geography}</b>')
    else:
        ticktext.append(geography)

fig.update_layout(yaxis=dict(tickmode='array', tickvals=df_plot['MSA'].unique(), ticktext=ticktext))

title = f'Annual Real Gross Regional Product Growth:  {year_min}-{year_max}'
fig.update_xaxes(tick0=0, dtick=1, ticksuffix='%', range = [0, 6])
fig.update_traces(hovertemplate='%{x}')
fig.update_layout(showlegend = False)

pt.plot_agol(fig, export, title, indicator, plot_name)


```
