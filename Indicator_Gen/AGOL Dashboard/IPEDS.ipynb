```python
# Set Indicator
indicator_name = 'Edu_2'
plot_name = 'by_field'
export = False


## Importing ---

file_name = f"{indicator_name}.xlsx"
df_mpo = pd.read_excel(os.path.join(path_plots, 'Data', file_name), sheet_name = 'AggregatedData')


## Organizing ---
df_plot = df_mpo.copy()

df_plot = df_plot.groupby(['year', 'Classification'], as_index = False)['grads'].sum()


df_plot['Sort'] = pd.Categorical(df_plot['Classification'], [
    'STEM', 'AHSS', 'Edu', 'Health', 'Business', 'Trades', 'Multi'
])
    
df_plot = df_plot.sort_values(['year', 'Sort'], ascending = [True, True])
df_plot = df_plot.drop(['Sort'], axis = 1)

df_plot.loc[df_plot['Classification'] == 'Edu', 'Classification'] = 'Education'

df_plot = df_plot.reset_index(drop = True)

display(df_plot.head())



## Plotting ---
color_map  = {
    'STEM':'#9DC209'
    , "AHSS":'#1E90FF'
    , "Education":'#1F45FC'
    , 'Health': '#7E587E'
    , 'Business': '#FBB117'
    , 'Trades': '#006A4E'
    , 'Multi':'#DC381F'
}


fig = px.line(df_plot, y='grads', x='year'
             , color='Classification'
             , color_discrete_map=color_map
             , markers = True)


title = '<b>Degrees Awarded in Sacramento Region Public Universities/Colleges, by Field</b>'
fig.update_yaxes(tick0=0, dtick=2000, tickformat = ',.0f', range=[0, 16100])
fig.update_xaxes(tick0=0, dtick=1, range = [2004.5, 2022.5])
fig.update_traces(hovertemplate='%{y}')


plot_agol(export=export)
```
