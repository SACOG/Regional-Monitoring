```python


import pandas as pd
from pathlib import Path
import plotly.express as px
from IPython.display import display
import sys
sys.path.append(str(Path.home() / 'Documents' / 'Projects' / 'Regional-Monitoring' / 'Indicator_Gen' / 'config'))
import plot as pt

PATH_PLOTS = Path(r"\\webmapping-svr\c$\inetpub\wwwroot\monitoring")
print('EXPORT Location: ' + str(PATH_PLOTS))
pd.options.display.float_format = '{:.2f}'.format

EXPORT=False
```

```python


indicator = 'Edu_2'
plot_name = 'degrees_awarded'

df_plot = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator}.xlsx", sheet_name='AggregatedData')
df_plot = df_plot.groupby(['year', 'level'], as_index = False)['grads'].sum()

df_plot['Sort'] = pd.Categorical(df_plot['level'], [
    'Associate'
    , "Bachelor"
    , "Graduate"
])

df_plot = df_plot.sort_values(['year', 'Sort'], ascending = [True, True]).drop(['Sort'], axis = 1).reset_index(drop = True)
display(df_plot.head())


fig = px.bar(df_plot, y='grads', x='year', color='level', color_discrete_map=pt.colors_edu)
title = '<b>Degrees Awarded in Sacramento Region Public Universities/Colleges, by Level</b>'
fig.update_yaxes(tick0=0, dtick=5000, tickformat = ',.0f', range=[0, 36000])
fig.update_xaxes(tick0=0, dtick=5)
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})

fig.add_annotation(
    x=0.9, y=-0.15,
    # text="*No data collected in 2021<br>due to covid pandemic",
    text=f"*Graduation year {df_plot['year'].max()} data is provisional",
    showarrow=False,
    xanchor="left", 
    yanchor="top",
    font=dict(size=10),
    xref="paper", yref="paper"
)


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```

```python


indicator = 'Edu_2'
plot_name = 'by_field'

df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator}.xlsx", sheet_name='AggregatedData')

df_plot = df_mpo.copy()
df_plot = df_plot.groupby(['year', 'Classification'], as_index = False)['grads'].sum()

df_plot = df_plot.sort_values(['year', 'grads'], ascending = [False, False]).reset_index(drop=True)
df_plot.loc[df_plot['Classification'] == 'Edu', 'Classification'] = 'Education'

display(df_plot.head())


fig = px.bar(df_plot, y='grads', x='year', color='Classification', color_discrete_map=pt.colors_edu)
title = '<b>Degrees Awarded in Sacramento Region Public Universities/Colleges, by Field</b>'
fig.update_yaxes(tick0=0, dtick=5000, tickformat = ',.0f', range=[0, 36000])
fig.update_xaxes(tick0=0, dtick=5, range = [df_plot['year'].min()-0.5, df_plot['year'].max()+0.5])
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})

fig.add_annotation(
    x=0.9, y=-0.15,
    # text="*No data collected in 2021<br>due to covid pandemic",
    text=f"*Graduation year {df_plot['year'].max()} data is provisional",
    showarrow=False,
    xanchor="left", 
    yanchor="top",
    font=dict(size=10),
    xref="paper", yref="paper"
)


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```
