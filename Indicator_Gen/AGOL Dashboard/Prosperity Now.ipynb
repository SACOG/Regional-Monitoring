```python
# Set Indicator
indicator_name = 'Income_5'
plot_name = 'negative_net_worth'
export = False


## Importing ---

file_name = f"Wealth_Metrics download.xlsx"
df_mpo = pd.read_excel(os.path.join(path_plots, 'Data', file_name), sheet_name = 'Data', skiprows=143)


## Organizing ---

df_plot = df_mpo.copy()

df_plot = df_plot[['Race', '0 Net Worth']]
df_plot.columns = ['Race_Ethnicity', 'Percentage']

df_plot['Percentage'] = round(df_plot['Percentage']*100, 1)


df_plot['Sort'] = pd.Categorical(df_plot['Race_Ethnicity'], [
    'Asian'
    , 'Black'
    , 'Hispanic'
    , 'Native'
    , 'Other'
    , 'White'
])
    
df_plot = df_plot.sort_values('Sort', ascending = True)
df_plot = df_plot.drop(['Sort'], axis = 1)


## Plotting ---


fig = px.bar(df_plot, x='Race_Ethnicity', y='Percentage')
fig.update_traces(marker_color='#FDBD01')


title = '<b>Sacramento Metro Region Households with Zero or Negative Net Worth by Race/Ethnicity</b>  <br><sup>4-County Sacramento MSA</sup> '
fig.update_yaxes(dtick=5, ticksuffix='%', range = [0,37])
fig.update_xaxes(dtick=1)
fig.update_traces(hovertemplate="%{y}")
fig.update_layout(showlegend=False)


plot_agol(export=export)
```
