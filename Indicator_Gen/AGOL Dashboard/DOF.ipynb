```python


EXPORT=False

import numpy as np
import pandas as pd
from pathlib import Path
import plotly.express as px
import sys
pd.options.display.float_format = '{:.2f}'.format

PATH_GIT     = Path.home()/'Documents'/'Projects'/'Regional-Monitoring'/'Indicator_Gen'
PATH_CONFIG0 = PATH_GIT/'config'

PATH_PLOTS = Path(r"\\webmapping-svr\c$\inetpub\wwwroot\monitoring")
print('Export Location: ' + str(PATH_PLOTS))


sys.path.append(str(PATH_CONFIG0))
import plot as pt

```

```python


indicator = 'Pop_1'
plot_name = 'counties'


df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} DOF Counties.xlsx", sheet_name='Data')

df_plot = df_mpo.copy()
df_plot['Sort'] = pd.Categorical(df_plot['County'], [
    'Sacramento'
    , 'Placer'
    , 'Yolo'
    , 'El Dorado'
    , 'Sutter'
    , 'Yuba'
])

df_plot = df_plot.sort_values(['Sort', 'Year'], ascending=[True, True])
df_plot = df_plot.drop(['Sort'], axis=1)
df_plot = df_plot.reset_index(drop = True)

display(df_plot.head())

color_map = {
    'Sacramento': '#1F45FC'
    , 'Placer': '#1E90FF'
    , 'Yolo': '#9DC209'
    , 'El Dorado': '#FBB117'
    , 'Sutter': '#DC381F'
    , 'Yuba': '#7E587E'
}


fig = px.bar(df_plot, x='Year', y='Population', color='County', color_discrete_map=color_map)


title = '<b>Total Population by County</b>  <br><sup>6-County Sacramento Region</sup>'
fig.update_yaxes(tick0=0, dtick=500000, range=[0, 3100000], tickformat=',.0f')
fig.update_xaxes(tick0=0, dtick=5)
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```

```python


indicator = 'Pop_5'
plot_name = 'GR_CA_regions'


df_mpo = pd.read_excel(PATH_PLOTS/'Data'/f"{indicator} DOF MPO.xlsx", sheet_name='Data')

df_plot = df_mpo.copy()

df_plot = df_plot[df_plot['MPO'] != 'Rest of CA']
df_plot['Population_GR'] = round(df_plot['Population_GR']*100, 1)

conditions = [
    df_plot['MPO'] == 'SACOG'
    , df_plot['MPO']  == 'CVCOG'
    , df_plot['MPO']  == 'MTC'
    , df_plot['MPO']  == 'SCAG'
    , df_plot['MPO']  == 'SANDAG'
]
choices = ['Sacramento', 'San Joaquin Valley', 'San Francisco Bay Area', 'Los Angeles', 'San Diego']
df_plot['Geography'] = np.select(conditions, choices, default='No')


df_plot['Sort'] = pd.Categorical(df_plot['Geography'], [
    'Sacramento'
    , 'San Joaquin Valley'
    , 'San Francisco Bay Area'
    , 'Los Angeles'
    , 'San Diego'
])

df_plot = df_plot.sort_values(['Sort', 'Year'], ascending=[True, True])
df_plot = df_plot.drop(['Sort'], axis=1)

df_plot = df_plot.reset_index(drop=True)

display(df_plot.head())



color_map = {
    'Sacramento':'#000000'
    , 'San Joaquin Valley':"#9DC209"
    , 'San Francisco Bay Area':"#DC381F"
    , 'Los Angeles':"#1E90FF"
    , 'San Diego':"#FBB117"
}

fig = px.line(df_plot, x='Year', y='Population_GR', color='Geography', color_discrete_map=color_map, markers=True)


title = '<b>Annual Population Growth, Sacramento Region and Other California Regions</b>'
fig.update_yaxes(dtick=0.5, ticksuffix='%', range = [-1.6,3.1])
fig.update_xaxes(dtick=5, range = [df_plot['Year'].min()-0.5, df_plot['Year'].max()+0.5])
fig.update_traces(hovertemplate="%{y}")
# fig.update_layout(legend=dict(orientation="h", yanchor="bottom", y=-0.15,xanchor="right", x=0.78))


pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```
