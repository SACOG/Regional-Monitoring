```python


import sys
from pathlib import Path

import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display

sys.path.append(str(Path.home() / 'Documents' / 'Projects' / 'Regional-Monitoring' / 'Indicator_Gen' / 'config'))
import plot as pt

PATH_DATA = Path(r"I:\Projects\Josh\Regional Monitoring\Employment Inventory\Data")
PATH_PLOTS = Path(r"\\webmapping-svr\c$\inetpub\wwwroot\monitoring")
print('EXPORT Location: ' + str(PATH_PLOTS))
pd.options.display.float_format = '{:.2f}'.format

def tidy_edd(df):
    df['Year/Quarter'] = df['Year'].astype(str) + ' Q' + df['Quarter'].astype(str)
    df['Sort'] = pd.Categorical(df['Year/Quarter'], ['2024 Q3', '2024 Q4', '2025 Q1', '2025 Q2', '2025 Q3'])
    df = df.sort_values(['Sort', 'Percent'], ascending=[True, False]).drop(['Sort'], axis=1).reset_index(drop=True)
    df['Percent'] = round(df['Percent']*100, 1)
    return df

EXPORT=False
```

```python

indicator = 'Jobs_6'
plot_name = 'sector'

df_plot = pd.read_excel(PATH_PLOTS / 'Data' / f"{indicator} Tradeable Sectors.xlsx", sheet_name='Region Tradeable')

df_plot = tidy_edd(df_plot)
display(df_plot.head())

fig = px.bar(df_plot, x='Year/Quarter', y='Percent', color='Tradeable', color_discrete_map=pt.colors_tradeable)
title = '<b>Percent of Regional Employment by Sector</b>  <br><sup>Six-County Sacramento Region</sup>'
fig.update_yaxes(tick0=0, ticksuffix = '%', range=[0, 101], dtick=25)
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})

pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```

```python


indicator = 'Jobs_6'
plot_name = 'tradable'

df_plot = pd.read_excel(PATH_PLOTS / 'Data' / f"{indicator} Tradeable Sectors.xlsx", sheet_name='Region Subsector')

df_plot = tidy_edd(df_plot)
display(df_plot.head())

fig = px.bar(df_plot, x='Year/Quarter', y='Number of Employees', color='Tradeable Subsector', color_discrete_map=pt.colors_tradeable_sub)
title = '<b>Estimated Regional Employment in Tradeable Subsectors</b>  <br><sup>Six-County Sacramento Region</sup>'
fig.update_yaxes(tick0=0, tickformat = ',.0f')
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})

pt.plot_agol(fig, EXPORT, title, indicator, plot_name)
```
