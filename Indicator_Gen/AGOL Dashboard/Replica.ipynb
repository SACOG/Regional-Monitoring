```python


import pandas as pd
from pathlib import Path
from IPython.display import display
import plotly.express as px
import sys
sys.path.append(str(Path.home() / 'Documents' / 'Projects' / 'Regional-Monitoring' / 'Indicator_Gen'/'config'))
import plot as pt

PATH_SP = Path(r'C:\Users\jfontes\Sacramento Area Council of Governments\Regional Monitoring and Reporting - Documents\Data\Next Gen of Mobility Solutions\Active Transportation\BikePed_1')
PATH_PLOTS = Path(r"\\webmapping-svr\c$\inetpub\wwwroot\monitoring")
print('Export Location: ' + str(PATH_PLOTS))

pd.options.display.float_format = '{:.2f}'.format


EXPORT=False



indicator='BikePed_1'
plot_name='trip_count'


file_bikeped = 'BikePed_1 Replica.xlsx'
sheet_name='County_Trend_Thu'
df=pd.read_excel(PATH_PLOTS/'Data'/file_bikeped, sheet_name=sheet_name,skiprows=1)


df = df.groupby(['Season', 'Mode'], as_index=False)['Trip Count'].sum()

df_fall   = df[df['Season'].str.contains('Fall'  )]
df_spring = df[df['Season'].str.contains('Spring')]

df_fall  ['Season'] = df_fall  ['Season'].str.replace(' Fall'  , '')
df_spring['Season'] = df_spring['Season'].str.replace(' Spring', '')

df_fall   = df_fall  .sort_values(['Season', 'Mode'], ascending=[True, False]).reset_index(drop=True)
df_spring = df_spring.sort_values(['Season', 'Mode'], ascending=[True, False]).reset_index(drop=True)


print('Fall Season')
display(df_fall)
print('\nSpring Season')
display(df_spring)


fig = px.bar(df_fall, x='Season', y='Trip Count', color='Mode', color_discrete_map=pt.colors_trips)
title = '<b>Estimated Number of Daily Active Transportation Trips</b> (Typical Thursday in the Fall Season) <br><sup>Six-County Sacramento Region</sup>'
fig.update_yaxes(tickformat = ',.0f')
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})

plot_name_fall=f'{plot_name}_fall'
pt.plot_agol(fig, EXPORT, title, indicator, plot_name_fall)


fig = px.bar(df_spring, x='Season', y='Trip Count', color='Mode', color_discrete_map=pt.colors_trips)
title = '<b>Estimated Number of Daily Active Transportation Trips</b> (Typical Thursday in the Spring Season) <br><sup>Six-County Sacramento Region</sup>'
fig.update_yaxes(tickformat = ',.0f')
fig.update_traces(hovertemplate='%{y}')
fig.update_layout(legend={'traceorder': 'reversed'})

plot_name_spring=f'{plot_name}_spring'
pt.plot_agol(fig, EXPORT, title, indicator, plot_name_spring)

```
