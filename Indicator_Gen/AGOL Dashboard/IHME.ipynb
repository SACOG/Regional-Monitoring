```python
# Set Indicator
indicator_name = 'Health_4'
plot_name = 'life_expectancy'
export = True


## Importing ---

file_name = f"{indicator_name} MPO IHME.xlsx"
df_mpo = pd.read_excel(os.path.join(path_plots, 'Data', file_name), sheet_name = 'MPO')



## Organizing ---

df_plot = df_mpo.copy()

df_plot = df_plot[df_plot['Race'] != 'Total']



conditions = [
    df_plot['Race'] == 'AIAN'
    , df_plot['Race']  == 'API'
    , df_plot['Race']  == 'Black'
    , df_plot['Race']  == 'Latino'
    , df_plot['Race']  == 'White'
]
choices = ['American Indian/Alaskan Native', 'Asian/Pacific Islander', 'Black or African American', 'Hispanic or Latino', 'White']
df_plot['Race_Ethnicity'] = np.select(conditions, choices, default = 'No')

df_plot = df_plot[df_plot['Race_Ethnicity'].isin(['American Indian/Alaskan Native', 'Asian/Pacific Islander', 'Black or African American', 'Hispanic or Latino', 'White'])]
display(df_plot.head())


## Plotting ---

color_map  = {
    'American Indian/Alaskan Native': '#7E587E'
    , 'Asian/Pacific Islander': '#9DC209'
    , 'Black or African American': '#1E90FF'
    , 'Hispanic or Latino': "#FBB117"
    , 'White': "#DC381F"
}

fig = px.line(df_plot, x='Year', y='Estimate', color='Race_Ethnicity', markers = True,
              color_discrete_map=color_map)

title = '<b>Life Expectancy by Race/Ethnicity</b>  <br><sup>6-County Sacramento Region</sup> '
fig.update_yaxes(dtick=5, range = [69, 91])
fig.update_xaxes(dtick=1, range = [1999.5, 2019.5])
fig.update_traces(hovertemplate="%{y}")


plot_agol(export=export)
```
